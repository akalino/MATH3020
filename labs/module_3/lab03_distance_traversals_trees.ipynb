# Lab 3: Distance, Traversals, and Trees

In this lab, we study three related themes in graph theory:

- the geometry of a graph through distance,
- traversal questions such as Euler and Hamilton problems,
- the structure of trees as minimally connected graphs.

These topics may seem different at first, but they are closely related.

Distance tells us how far apart vertices are.
Traversal asks how we can move through a graph.
Trees provide especially clean examples of both ideas.

### Part 9: Characterizations of trees

Trees can be recognized in several equivalent ways.

We now explore some of the most important characterizations:

- connected and acyclic,

- minimally connected.

```python
n = T.number_of_nodes()
m = T.number_of_edges()

print("n =", n)
print("m =", m)
print("m = n - 1:", m == n - 1)
```
